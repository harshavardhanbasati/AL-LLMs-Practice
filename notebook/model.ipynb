```python
import pandas as pd
data = {
     "hours_studied": [2, 3, 4, 5, 6, 7, 8, 9, 10],
    "attendance": [60, 65, 70, 75, 80, 85, 90, 92, 95],
    "previous_score": [45, 50, 55, 60, 65, 70, 75, 80, 85],
    "assignments": [4, 5, 5, 6, 7, 8, 8, 9, 10],
    "sleep_hours": [5, 5, 6, 6, 7, 7, 8, 8, 8],
    "final_score": [48, 52, 58, 63, 69, 74, 79, 84, 89]
}

df = pd.DataFrame(data)

print(df)
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["hours_studied"], df["final_score"])

plt.xlabel("Hours Studied")
plt.ylabel("Final Score")
plt.title("Hours Studied vs Final Score")

plt.show()
```

```python
X = df[
    [
        "hours_studied",
        "attendance",
        "previous_score",
        "assignments",
        "sleep_hours"
    ]
]

Y = df["final_score"]


from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size = 0.2,
    random_state = 42
)
print("Training data", X_train.shape)
print("Testing data",X_test.shape)

print(df.columns)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train,Y_train)

predictions = model.predict(X_test)
print(predictions)
print("Actual:",Y_test.values)
print("predicted:", predictions)
```

```python
from sklearn.metrics import r2_score

score = r2_score(Y_test, predictions)

print("R^2 Score: ", score)
```
